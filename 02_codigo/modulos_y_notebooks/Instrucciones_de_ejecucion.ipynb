## 3. Los controles de la animación

| Control | Qué hace |
| **Iniciar** | Arranca la reproducción automática, paso a paso, hasta llegar a la meta o agotar la frontera |
| **Pausar** | Detiene la reproducción automática donde vaya |
| **Avanzar** | Avanza un solo paso |
| **Reiniciar** | Vuelve al paso 0 |
| **Velocidad (ms)** | Control deslizante: milisegundos de espera entre paso y paso |
