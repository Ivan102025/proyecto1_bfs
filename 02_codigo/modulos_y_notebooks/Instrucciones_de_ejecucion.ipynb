## 1. Qué hay que instalar
**Nada.** Google Colab ya trae instalado todo lo que usa el proyecto:

## 2. Cómo abrir y ejecutar el notebook principal

1. Subir `Proyecto1_BFS_Colab.ipynb` a Google Colab (Archivo → Subir
   notebook, o abrirlo directamente desde Google Drive).
2. Ejecutar las celdas **en orden, de arriba hacia abajo**, con
   `Shift + Enter` en cada una, o con el menú **Entorno de ejecución →
   Ejecutar todas**. No se debe saltar ninguna celda ni ejecutarlas en
   otro orden: las funciones se definen en una celda y se usan en las
   celdas siguientes.
3. Las celdas de animación (sección 6 en adelante) tardan unos segundos
   en generar el reproductor, porque arman un cuadro por cada paso de
   BFS. Es normal.

## 5. Problemas conocidos
- Los mapas y el CSV de métricas se guardan solo en el almacenamiento
  temporal de la sesión de Colab (`mapas/` y `resultados/`); se pierden
  si el entorno de ejecución se reinicia. Para conservarlos hay que
  descargarlos desde el panel de archivos de la izquierda.
- La visualización usa `matplotlib` en vez de Tkinter, porque Colab no
  tiene entorno gráfico de escritorio donde abrir una ventana.
